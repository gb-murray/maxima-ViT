```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set a clean style for the plots
sns.set_theme(style="whitegrid")

# Load the training logs (replace filenames with your actual paths)
# Expected columns: ['step', 'train', 'test']
phase1_df = pd.read_csv("")
phase2_df = pd.read_csv("")
```

```python
# Create a figure with 2 paired subplots side-by-side
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)
# fig.suptitle('Model Loss Across Training Phases', fontsize=14, y=1.05)

# ---------------------------------------------------------
# Plot 1: Phase 1 (Learning Rate changes at step 20)
# ---------------------------------------------------------
axes[0].plot(phase1_df['Step'], phase1_df['Train'], label='Train', color='#1f77b4', linewidth=2)
axes[0].plot(phase1_df['Step'], phase1_df['Test'], label='Test', color='#ff7f0e', linewidth=2, alpha=0.8)

axes[0].set_title('(a)')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].legend(loc='upper right')
axes[0].set_xlim(0, phase1_df['Step'].max()+1)

# Add vertical line at step 20 to mark the LR change
axes[0].axvline(x=20, color='gray', linestyle='--', alpha=0.7)

# Annotate LR before and after the step change
# Using transAxes keeps the text perfectly placed regardless of the data scale
bbox_props = dict(boxstyle="round,pad=0.3", fc="white", ec="gray", alpha=1.0)
axes[0].text(0.125, 0.75, 'LR = 1e-4 \n Head Layers \n Only', transform=axes[0].transAxes, 
             fontsize=10, ha='center',bbox=bbox_props)
axes[0].text(0.30, 0.79, 'LR = 2e-5 \n All Layers', transform=axes[0].transAxes, 
             fontsize=10, ha='center', bbox=bbox_props)

# ---------------------------------------------------------
# Plot 2: Phase 2 (Constant Learning Rate)
# ---------------------------------------------------------
axes[1].plot(phase2_df['Step'], phase2_df['Train'], label='Train', color='#1f77b4', linewidth=2)
axes[1].plot(phase2_df['Step'], phase2_df['Test'], label='Test', color='#ff7f0e', linewidth=2, alpha=0.8)

axes[1].set_title('(b)')
axes[1].set_xlabel('Epoch')
axes[1].set_xlim(0, phase2_df['Step'].max()+1)
axes[1].legend(loc='upper right')

# Annotate the constant LR for consistency
axes[1].text(0.30, 0.83, 'LR = 1e-6', transform=axes[1].transAxes, 
             fontsize=10, ha='center', bbox=bbox_props)

# Adjust layout so nothing overlaps
plt.tight_layout()
plt.show()
```
