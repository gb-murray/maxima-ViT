```python
import pyFAI.calibrant
from pyFAI.integrator.azimuthal import AzimuthalIntegrator
from pyFAI.detectors import detector_factory
import os
from matplotlib.colors import SymLogNorm
import matplotlib.pyplot as plt
```

```python
cal = pyFAI.calibrant.get_calibrant("alpha_Al2O3")
output_filename = "custom_alpha_Al2O3.cel"

with open(output_filename, "w") as f:
    f.write("# Calibrant: Custom Alpha Al2O3 with Relative Intensities\n")
    f.write("# Cell: Placeholder\n")
    f.write("# d_spacing  # (h k l)  mult intensity\n")
    
    for i, d in enumerate(cal.dSpacing):
        f.write(f"{d:.8f} # ({i+1} 0 0)    1  1.0\n")

print(f"Success! Created '{output_filename}' at: {os.getcwd()}")
```

```python
custom_calibrant = pyFAI.calibrant.Calibrant("custom_alpha_Al2O3.cel")

ai = AzimuthalIntegrator(
    dist = 0.12,
    poni1 = 0.04,
    poni2 = 0.04,
    rot1 = 0,
    rot2 = 0,
    rot3 = 0,
    wavelength = 0.5121197787410171e-10,
    detector = "Eiger2Cdte_1M"
)

pattern = custom_calibrant.fake_calibration_image(ai)

plt.imshow(pattern, cmap="viridis", origin="lower", norm=SymLogNorm(linthresh=1))
```

```python
q, intensity = ai.integrate1d(pattern, 1000, unit="2th_deg", correctSolidAngle=False)
plt.plot(q, intensity)
plt.xlabel("2θ (deg)")
plt.ylabel("Intensity")
#plt.xlim(18, 20)
plt.show()
```
